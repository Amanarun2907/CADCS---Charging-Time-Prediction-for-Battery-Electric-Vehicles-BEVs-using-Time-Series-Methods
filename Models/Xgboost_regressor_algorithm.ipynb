```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

### Reading the files from the data folder : X_train , X_test , y_train and y_test

```python
X_train = pd.read_csv(r"../data/X_train.csv")
X_test = pd.read_csv(r"../data/X_test.csv")
y_train = pd.read_csv(r"../data/y_train.csv")
y_test = pd.read_csv(r"../data/y_test.csv")
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.linear_model import LinearRegression, Ridge,Lasso
from sklearn.neighbors import KNeighborsRegressor
from xgboost import XGBRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
```

### Creating a function to calculate the mean squared error of the model , mean absolute error , root mean square error and r square value 

```python
##Create a Function to Evaluate Model
def evaluate_model(true, predicted):
    mae = mean_absolute_error(true, predicted)
    mse = mean_squared_error(true, predicted)
    rmse = np.sqrt(mean_squared_error(true, predicted))
    r2_square = r2_score(true, predicted)
    return mae, rmse, r2_square
```

### Building an Adaboost Regressor Model for predicting the charging time of a BEV.

```python
## Beginning Model Training
models = {
    "Linear Regression": LinearRegression(),
    "Lasso": Lasso(),
    "Ridge": Ridge(),
    "K-Neighbors Regressor": KNeighborsRegressor(),
    "Decision Tree": DecisionTreeRegressor(),
    "Random Forest Regressor": RandomForestRegressor(),
    "Adaboost Regressor":AdaBoostRegressor(),
    "Graident BoostRegressor":GradientBoostingRegressor(),
    "Xgboost Regressor":XGBRegressor()
   
}

for i in range(len(list(models))):
    model = list(models.values())[i]
    model.fit(X_train, y_train) # Train model

    # Make predictions
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    
    # Evaluate Train and Test dataset
    model_train_mae , model_train_rmse, model_train_r2 = evaluate_model(y_train, y_train_pred)

    model_test_mae , model_test_rmse, model_test_r2 = evaluate_model(y_test, y_test_pred)

    
    print(list(models.keys())[i])
    
    print('Model performance for Training set')
    print("- Root Mean Squared Error: {:.4f}".format(model_train_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_train_mae))
    print("- R2 Score: {:.4f}".format(model_train_r2))

    print('----------------------------------')
    
    print('Model performance for Test set')
    print("- Root Mean Squared Error: {:.4f}".format(model_test_rmse))
    print("- Mean Absolute Error: {:.4f}".format(model_test_mae))
    print("- R2 Score: {:.4f}".format(model_test_r2))
    
    print('='*35)
    print('\n')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# --- 1. Load the Dataset (assuming preprocessed data is available) ---
# Ensure X_train.csv, X_test.csv, y_train.csv, y_test.csv are accessible.
try:
    X_train = pd.read_csv(r"../data/X_train.csv")
    X_test = pd.read_csv(r"../data/X_test.csv")
    y_train = pd.read_csv(r"../data/y_train.csv")
    y_test = pd.read_csv(r"../data/y_test.csv")
except FileNotFoundError:
    print("Error: One or more data files (X_train.csv, X_test.csv, y_train.csv, y_test.csv) not found.")
    print("Please ensure your preprocessed and split data files are in the correct directory.")
    exit()

# Flatten y_train and y_test to 1D arrays as expected by scikit-learn models
# This addresses the DataConversionWarning observed in your notebook.
y_train_flat = y_train.values.ravel()
y_test_flat = y_test.values.ravel()

# --- 2. Initialize and Train Linear Regression Model ---
linear_reg_model = LinearRegression()
linear_reg_model.fit(X_train, y_train_flat) # Train model on flattened y_train

# --- 3. Make Predictions on the Test Set ---
y_test_pred_lr = linear_reg_model.predict(X_test)

# --- 4. Calculate Evaluation Metrics for Linear Regression (Optional, but good for context) ---
# You already have an evaluate_model function in your notebook, but including here for self-containment.
mae_lr = mean_absolute_error(y_test_flat, y_test_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y_test_flat, y_test_pred_lr))
r2_lr = r2_score(y_test_flat, y_test_pred_lr)

print(f"--- Linear Regression Model Performance (Test Set) ---")
print(f"Root Mean Squared Error (RMSE): {rmse_lr:.4f}")
print(f"Mean Absolute Error (MAE): {mae_lr:.4f}")
print(f"R2 Score: {r2_lr:.4f}\n")


# --- Plot 1: Actual vs. Predicted Values ---
plt.figure(figsize=(10, 8))
sns.scatterplot(x=y_test_flat, y=y_test_pred_lr, alpha=0.6, color='blue')
plt.plot([y_test_flat.min(), y_test_flat.max()], [y_test_flat.min(), y_test_flat.max()],
         '--r', linewidth=2, label='Perfect Prediction Line') # Ideal line (y=x)
plt.title('Linear Regression: Actual vs. Predicted Charging Duration (Test Set)', fontsize=16)
plt.xlabel('Actual Charging Duration (minutes)', fontsize=12)
plt.ylabel('Predicted Charging Duration (minutes)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- Plot 2: Residuals Plot ---
# Residuals = Actual Values - Predicted Values
residuals_lr = y_test_flat - y_test_pred_lr

plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test_pred_lr, y=residuals_lr, alpha=0.6, color='lightcoral')
plt.axhline(y=0, color='r', linestyle='--', linewidth=2, label='Zero Residual Line') # Line at y=0
plt.title('Linear Regression: Residuals Plot (Test Set)', fontsize=16)
plt.xlabel('Predicted Charging Duration (minutes)', fontsize=12)
plt.ylabel('Residuals (Actual - Predicted)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

print("Two plots for Linear Regression evaluation have been generated: Actual vs. Predicted, and Residuals Plot.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import AdaBoostRegressor # Import AdaBoostRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.tree import DecisionTreeRegressor # AdaBoost often uses a DecisionTree as base estimator

# --- 1. Load the Dataset (assuming preprocessed data is available) ---
# Ensure X_train.csv, X_test.csv, y_train.csv, y_test.csv are accessible.
try:
    X_train = pd.read_csv(r"../data/X_train.csv")
    X_test = pd.read_csv(r"../data/X_test.csv")
    y_train = pd.read_csv(r"../data/y_train.csv")
    y_test = pd.read_csv(r"../data/y_test.csv")
    
except FileNotFoundError:
    print("Error: One or more data files (X_train.csv, X_test.csv, y_train.csv, y_test.csv) not found.")
    print("Please ensure your preprocessed and split data files are in the correct directory.")
    exit()

# Flatten y_train and y_test to 1D arrays as expected by scikit-learn models
# This addresses the DataConversionWarning observed in your notebook.
y_train_flat = y_train.values.ravel()
y_test_flat = y_test.values.ravel()

# --- 2. Initialize and Train AdaBoost Regressor Model ---
# Using a DecisionTreeRegressor as the base estimator, which is common for AdaBoost.
# You can adjust parameters like n_estimators, learning_rate if needed.
adaboost_model = AdaBoostRegressor(estimator=DecisionTreeRegressor(max_depth=4),
                                   n_estimators=100, # Number of boosting stages
                                   learning_rate=1.0, # Contribution of each estimator
                                   random_state=42) # For reproducibility

adaboost_model.fit(X_train, y_train_flat) # Train model on flattened y_train

# --- 3. Make Predictions on the Test Set ---
y_test_pred_ab = adaboost_model.predict(X_test)

# --- 4. Calculate Evaluation Metrics for AdaBoost (Optional, but good for context) ---
mae_ab = mean_absolute_error(y_test_flat, y_test_pred_ab)
rmse_ab = np.sqrt(mean_squared_error(y_test_flat, y_test_pred_ab))
r2_ab = r2_score(y_test_flat, y_test_pred_ab)

print(f"--- AdaBoost Regressor Model Performance (Test Set) ---")
print(f"Root Mean Squared Error (RMSE): {rmse_ab:.4f}")
print(f"Mean Absolute Error (MAE): {mae_ab:.4f}")
print(f"R2 Score: {r2_ab:.4f}\n")


# --- Plot 1: Actual vs. Predicted Values for AdaBoost ---
plt.figure(figsize=(10, 8))
sns.scatterplot(x=y_test_flat, y=y_test_pred_ab, alpha=0.6, color='mediumseagreen')
plt.plot([y_test_flat.min(), y_test_flat.max()], [y_test_flat.min(), y_test_flat.max()],
         '--r', linewidth=2, label='Perfect Prediction Line') # Ideal line (y=x)
plt.title('AdaBoost Regressor: Actual vs. Predicted Charging Duration (Test Set)', fontsize=16)
plt.xlabel('Actual Charging Duration (minutes)', fontsize=12)
plt.ylabel('Predicted Charging Duration (minutes)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- Plot 2: Residuals Plot for AdaBoost ---
# Residuals = Actual Values - Predicted Values
residuals_ab = y_test_flat - y_test_pred_ab

plt.figure(figsize=(10, 6))
sns.scatterplot(x=y_test_pred_ab, y=residuals_ab, alpha=0.6, color='salmon')
plt.axhline(y=0, color='r', linestyle='--', linewidth=2, label='Zero Residual Line') # Line at y=0
plt.title('AdaBoost Regressor: Residuals Plot (Test Set)', fontsize=16)
plt.xlabel('Predicted Charging Duration (minutes)', fontsize=12)
plt.ylabel('Residuals (Actual - Predicted)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

print("Two plots for AdaBoost Regressor evaluation have been generated: Actual vs. Predicted, and Residuals Plot.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import AdaBoostRegressor
from xgboost import XGBRegressor
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from sklearn.tree import DecisionTreeRegressor # For AdaBoost base estimator

# --- 1. Load the Dataset ---
# Ensure X_train.csv, X_test.csv, y_train.csv, y_test.csv are accessible.
try:
    X_train = pd.read_csv(r"../data/X_train.csv")
    X_test = pd.read_csv(r"../data/X_test.csv")
    y_train = pd.read_csv(r"../data/y_train.csv")
    y_test = pd.read_csv(r"../data/y_test.csv")
    
except FileNotFoundError:
    print("Error: One or more data files (X_train.csv, X_test.csv, y_train.csv, y_test.csv) not found.")
    print("Please ensure your preprocessed and split data files are in the correct directory.")
    exit()

# Flatten y_train and y_test to 1D arrays
y_train_flat = y_train.values.ravel()
y_test_flat = y_test.values.ravel()

# --- 2. Train and Evaluate Each Model to Get R2 Scores ---

# Linear Regression
linear_reg_model = LinearRegression()
linear_reg_model.fit(X_train, y_train_flat)
y_test_pred_lr = linear_reg_model.predict(X_test)
r2_lr = r2_score(y_test_flat, y_test_pred_lr)

# AdaBoost Regressor
# Using the same base estimator and parameters as previously discussed for consistency
adaboost_model = AdaBoostRegressor(estimator=DecisionTreeRegressor(max_depth=4),
                                   n_estimators=100,
                                   learning_rate=1.0,
                                   random_state=42)
adaboost_model.fit(X_train, y_train_flat)
y_test_pred_ab = adaboost_model.predict(X_test)
r2_ab = r2_score(y_test_flat, y_test_pred_ab)

# XGBoost Regressor
# Using the same parameters as previously discussed for consistency
xgboost_model = XGBRegressor(objective='reg:squarederror',
                             n_estimators=100,
                             learning_rate=0.1,
                             max_depth=5,
                             random_state=42)
xgboost_model.fit(X_train, y_train_flat)
y_test_pred_xgb = xgboost_model.predict(X_test)
r2_xgb = r2_score(y_test_flat, y_test_pred_xgb)

# --- 3. Collect R2 Scores for Plotting ---
# Ensure the order here matches the order you want on the plot if not sorting
model_names = ['Linear Regression', 'AdaBoost Regressor', 'XGBoost Regressor']
r2_scores = [r2_lr, r2_ab, r2_xgb]

# Create a DataFrame for easy plotting
r2_df = pd.DataFrame({
    'Model': model_names,
    'R2 Score': r2_scores
})

# Sort by R2 Score for better visualization, this will reorder the 'Model' column
r2_df = r2_df.sort_values(by='R2 Score', ascending=False).reset_index(drop=True)

# --- 4. Create the R2 Score Comparison Plot ---
plt.figure(figsize=(10, 7))
# Use the sorted DataFrame directly for plotting
sns.barplot(x='Model', y='R2 Score', data=r2_df, palette='viridis')

# Add R2 score values on top of the bars
# Iterate through the rows of the *sorted* DataFrame for correct placement
for index, row in r2_df.iterrows():
    plt.text(index, row['R2 Score'] + 0.005, f"{row['R2 Score']:.4f}",
             color='black', ha="center", va='bottom', fontsize=10)

plt.ylim(0, 1.05) # R2 score ranges from -infinity to 1, but typically 0-1 for good models
plt.title('Comparison of R2 Scores Across Regression Models (Test Set)', fontsize=16)
plt.xlabel('Regression Model', fontsize=12)
plt.ylabel('R2 Score', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

print("A bar plot comparing the R2 scores of Linear Regression, AdaBoost, and XGBoost models has been generated with corrected labels.")
```
