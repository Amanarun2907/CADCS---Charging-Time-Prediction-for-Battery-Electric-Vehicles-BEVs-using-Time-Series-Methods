## Charging Time Prediction for Battery Electric Vehicles (BEVs) using Time Series Methods

## Project Overview
### Battery Electric Vehicles (BEVs) play a significant role in reducing energy consumption and air pollution, offering a cleaner alternative to conventional internal combustion engine vehicles. Despite their advantages, BEVs face challenges related to limited driving range and prolonged charging durations. These issues contribute to range anxiety, which significantly affects user adoption and satisfaction. Predicting charging time accurately based on real-world data can empower drivers with better travel planning and reduce anxiety. This project aims to develop a predictive model that improves the estimation of BEV charging times by leveraging actual operational data from BEVs

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import numpy as np

# --- 1. Load the Dataset ---
# Ensure 'Dataset_BEV_CADCS_Project.csv' is accessible in the environment.
try:
    df = pd.read_csv(r"../data/cadcs_live_project_clean_data.csv")
except FileNotFoundError:
    print("Error: 'Dataset_BEV_CADCS_Project.csv' not found. Please ensure the file is uploaded.")
    exit()

# --- 2. Define Numerical Features for Demonstration ---
# Selecting a few key numerical features that typically have different scales.
# These features are chosen to clearly illustrate the effect of standardization.
num_features_to_demonstrate = [
    'energy_delivered_kwh',
    'battery_capacity_kwh',
    'ambient_temperature_c',
    'charger_power_kw',
    'total_cost_inr',
    'queue_wait_time_min'
]

# Filter to ensure only existing features are used
existing_features = [f for f in num_features_to_demonstrate if f in df.columns]
if not existing_features:
    print("Error: None of the selected features for demonstration exist in the dataset.")
    exit()

# Handle potential missing values in the demonstration features by filling with median
# This is done to prevent plotting errors for this specific visualization.
# In a full preprocessing pipeline, more robust missing value strategies would be applied.
for feature in existing_features:
    if df[feature].isnull().any():
        df[feature].fillna(df[feature].median(), inplace=True)
        # print(f"Filled missing values in '{feature}' with its median for demonstration.")


# --- Plot 1: Distributions BEFORE Standardization ---
# Create a figure to hold subplots for original distributions
plt.figure(figsize=(16, 10)) # Adjust figure size for better readability
plt.suptitle('Feature Distributions BEFORE Standardization', fontsize=18, y=1.02)

# Determine grid size dynamically based on number of features
n_features = len(existing_features)
n_rows = (n_features + 2) // 3 # Calculate rows needed for max 3 columns
n_cols = 3

for i, feature in enumerate(existing_features):
    plt.subplot(n_rows, n_cols, i + 1) # Create subplot for each feature
    sns.histplot(df[feature], kde=True, bins=30, color='skyblue', edgecolor='black')
    plt.title(f'{feature}', fontsize=14)
    plt.xlabel('Original Value', fontsize=12)
    plt.ylabel('Count', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap
plt.show()


# --- Apply StandardScaler ---
scaler = StandardScaler()
# Fit and transform only the selected numerical features
df_scaled_features = scaler.fit_transform(df[existing_features])

# Convert the scaled array back to a DataFrame for easier plotting
df_scaled = pd.DataFrame(df_scaled_features, columns=existing_features)


# --- Plot 2: Distributions AFTER Standardization ---
# Create a new figure to hold subplots for scaled distributions
plt.figure(figsize=(16, 10)) # Same figure size for consistent comparison
plt.suptitle('Feature Distributions AFTER Standardization (Scaled)', fontsize=18, y=1.02)

for i, feature in enumerate(existing_features):
    plt.subplot(n_rows, n_cols, i + 1) # Create subplot for each feature
    sns.histplot(df_scaled[feature], kde=True, bins=30, color='lightcoral', edgecolor='black')
    plt.title(f'{feature}', fontsize=14)
    plt.xlabel('Scaled Value (Mean=0, Std Dev=1)', fontsize=12)
    plt.ylabel('Count', fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout
plt.show()

print("\nTwo plots have been generated, illustrating the distributions before and after standardization.")
print("The first plot shows features with their original scales, while the second shows them transformed to have a mean of 0 and a standard deviation of 1.")
```
